##### Identify Price Outliers Using IQR Method 🇮🇳

```python
Q1 = df['Price'].quantile(0.25)
Q3 = df['Price'].quantile(0.75)

iqr = Q3 - Q1

lower_bound = Q1 - 1.5 * iqr
upper_bound = Q3 + 1.5 * iqr

print('Q1:', Q1)
print('Q3:', Q3)
print('IQR:', iqr)
print('lower bound:', lower_bound)
print('upper bound:', upper_bound)
```

```python
outliers = df[
    (df['Price'] < lower_bound) |
    (df['Price'] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

##### See the Actual Outlier Rows

```python
outliers[['Price', 'living area', 'grade of the house',
          'number of bathrooms', 'number of bedrooms']].sort_values(
    by='Price',
    ascending=False
).head(10)
```

##### Handling id and Date Columns

```python
df = df.drop('id', axis=1)

df['Date'] = pd.to_datetime(
    df['Date'],
    origin='1899-12-30',
    unit='D'
)

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

df = df.drop('Date', axis=1)

df.head()
```

##### Train/Test Split

```python
x = df.drop('Price', axis=1)
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,
    random_state=42
)
```

```python
print("X_train:", x_train.shape)
print("X_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

#### LinearRegression

```python
# import LinearRegression
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()
```

```python
lr_model.fit(x_train,y_train)
```

```python
# Prediction
y_pred = lr_model.predict(x_test)
print(y_pred[:10])
```

##### Evaluate the Linear Regression Model Using MAE, RMSE, and R²

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

#### Decision Tree Regression

```python
# import Decision Tree Regression
from sklearn.tree import DecisionTreeRegressor

dt_model = DecisionTreeRegressor(
    random_state=42
)
dt_model.fit(x_train,y_train)
```

```python
# Predict
dt_pred = dt_model.predict(x_test)
print(dt_pred)
```

##### Evaluate the Decision Tree Model Using MAE, RMSE, and R²

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score
import numpy as np

dt_mae = mean_absolute_error(y_test,dt_pred)

dt_rmse = np.sqrt(mean_squared_error(y_test, dt_pred))

dt_r2 = r2_score(y_test,dt_pred)

# Decision Tree Result Check
print("Decision Tree MAE:", dt_mae)
print("Decision Tree RMSE:", dt_rmse)
print("Decision Tree R2:", dt_r2)
```

#### Random Forest Regression

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=1
)

rf_model.fit(x_train,y_train)
```

```python
# Predict
rf_pred = rf_model.predict(x_test)
print(rf_pred)
```

```python
# evaluate
rf_mae = mean_absolute_error(y_test, rf_pred)

rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))

rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest MAE:", rf_mae)
print("Random Forest RMSE:", rf_rmse)
print("Random Forest R2:", rf_r2)
```

##### Models Comparison

```python
feature_importance = pd.DataFrame({
    'Feature': x_train.columns,
    'Importance': rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

feature_importance.head(10)
```

```python
top_features = feature_importance.head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_features,
    x='Importance',
    y='Feature'
)

plt.title('Top 10 Feature Importance')
plt.xlabel('Importance')
plt.ylabel('Feature')

plt.show()
```

##### Random Forest Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV

rf_param_grid = {
    'n_estimators': [50, 100],
    'max_depth': [10, 15],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}
```

```python
# gridsearchch
rf_grid = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42, n_jobs=-1),
    param_grid=rf_param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1
)


```

```python
rf_grid.fit(x_train,y_train)
```

```python
print('best parameters')
print(rf_grid.best_params_)

print("Best CV R2:")
print(rf_grid.best_score_)
```

#### Select the Best Random Forest Model

```python
# find best data
best_rf_model = rf_grid.best_estimator_
```

```python
# predction on test data
rf_tuned_pred = best_rf_model.predict(x_test)
```

```python
# Performance check
tuned_rf_mae = mean_absolute_error(y_test,rf_tuned_pred)

tuned_rf_rmse = np.sqrt(
    mean_squared_error(y_test,rf_tuned_pred)
)

tuned_rf_r2 = r2_score(y_test, rf_tuned_pred)

```

```python
tuned_rf_mae = mean_absolute_error(y_test, rf_tuned_pred)
tuned_rf_rmse = np.sqrt(mean_squared_error(y_test, rf_tuned_pred))
tuned_rf_r2 = r2_score(y_test, rf_tuned_pred)
```

```python
# Create Linear Regression metric variables
lr_mae = mean_absolute_error(y_test, y_pred)
lr_rmse = np.sqrt(mean_squared_error(y_test, y_pred))
lr_r2 = r2_score(y_test, y_pred)
```

##### Compare All Models

```python
model_comparison = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Decision Tree',
        'Random Forest',
        'Tuned Random Forest'
    ],
    'MAE': [
        lr_mae,
        dt_mae,
        rf_mae,
        tuned_rf_mae
    ],
    'RMSE': [
        lr_rmse,
        dt_rmse,
        rf_rmse,
        tuned_rf_rmse
    ],
    'R2 Score': [
        lr_r2,
        dt_r2,
        rf_r2,
        tuned_rf_r2
    ]
})

print(model_comparison)
```
